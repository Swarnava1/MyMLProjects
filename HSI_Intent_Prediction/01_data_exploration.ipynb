```python
import urllib.request
import zipfile
import pandas as pd
import os

url = "https://archive.ics.uci.edu/static/public/240/human+activity+recognition+using+smartphones.zip"
zip_filename = "har_dataset.zip"
extract_folder = "har_data"

# 1. Download the raw zip file from the server
print("1. Downloading raw zip file...")
urllib.request.urlretrieve(url, zip_filename)

# 2. Extract the contents into a new folder called 'har_data'
print("2. Extracting files...")
with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
    zip_ref.extractall(extract_folder)

# 3. Load the data using Pandas
print("3. Loading sensor data into tables...")
# The researchers split the data into "train" (for learning) and "test" (for evaluating).
# We will load the training data first.
X_train_path = os.path.join(extract_folder, "UCI HAR Dataset", "train", "X_train.txt")
y_train_path = os.path.join(extract_folder, "UCI HAR Dataset", "train", "y_train.txt")

# We use sep='\s+' which is a Regular Expression telling Pandas: 
# "Treat ANY amount of whitespace as a column separator"
X = pd.read_csv(X_train_path, sep='\s+', header=None)
y = pd.read_csv(y_train_path, sep='\s+', header=None, names=['Activity_Label'])

# 4. Verify the shape
print(f"\nSuccess!")
print(f"Features (X) Shape: {X.shape}")
print(f"Targets (y) Shape: {y.shape}")

display(X.head())
```
